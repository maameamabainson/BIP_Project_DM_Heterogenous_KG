For node2vec & r-gcn, switching node name and node index so that nodes will be integers

```python
import pandas as pd
import networkx as nx
import numpy as np
import random

RANDOM_STATE = 0
random.seed(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
```

```python
full_L1 = pd.read_csv(f'kg_data/full_L1.csv')
l1_nodes = pd.read_csv(f'kg_data/l1_nodes.csv')

full_L2 = pd.read_csv(f'kg_data/full_L2.csv')
l2_nodes = pd.read_csv(f'kg_data/l2_nodes.csv')

full_L3 = pd.read_csv(f'kg_data/full_L3.csv')
l3_nodes = pd.read_csv(f'kg_data/l3_nodes.csv')

whole_kg_data = pd.read_csv(f'kg_data/whole_kg_data.csv')
whole_nodes = pd.read_csv(f'kg_data/whole_nodes.csv')
```

```python
#Function for creating list of nodes and edges based on networkx documentation for adding nodes and edges, emphasis on relation and node types for rgcn

def for_rgcn_create_node_list(nodesdf):
    node_list = []
    for i in range(len(nodesdf)):
        node_list.append((int(nodesdf.loc[i,'node_index']),{'node_name':nodesdf.loc[i,'node_name'],'node_type_index':nodesdf.loc[i,'node_type_index']}))
    
    return node_list

def for_rgcn_create_relation_list(edgesdf):
    edge_list = []

    for i in range(len(edgesdf)):
        edge_list.append((int(edgesdf.loc[i,'x_index']),int(edgesdf.loc[i,'y_index']),{'relation_type_index': edgesdf.loc[i,'relation_type_index']}))
    
    return edge_list
```

```python
#Function for creating list of nodes and edges based on networkx documentation for adding nodes and edges, for homogenous n2v

def for_n2v_create_node_list(nodesdf):
    node_list = []
    for i in range(len(nodesdf)):
        node_list.append((int(nodesdf.loc[i,'node_index']),{'node_name':nodesdf.loc[i,'node_name'],'node_type_index':nodesdf.loc[i,'node_type_index']}))
    
    return node_list

def for_n2v_create_relation_list(edgesdf):
    edge_list = []

    for i in range(len(edgesdf)):
        edge_list.append((int(edgesdf.loc[i,'x_index']),int(edgesdf.loc[i,'y_index']),{'relation': edgesdf.loc[i,'relation']}))
    
    return edge_list
```

```python
n2v_host_layerkg = nx.DiGraph()
n2v_host_layerkg.add_nodes_from(for_n2v_create_node_list(l1_nodes))
n2v_host_layerkg.add_edges_from(for_n2v_create_relation_list(full_L1))

n2v_causal_layerkg = nx.DiGraph()
n2v_causal_layerkg.add_nodes_from(for_n2v_create_node_list(l2_nodes))
n2v_causal_layerkg.add_edges_from(for_n2v_create_relation_list(full_L2))

n2v_amr_layerkg = nx.DiGraph()
n2v_amr_layerkg.add_nodes_from(for_n2v_create_node_list(l3_nodes))
n2v_amr_layerkg.add_edges_from(for_n2v_create_relation_list(full_L3))

n2v_wholekg = nx.DiGraph()
n2v_wholekg.add_nodes_from(for_n2v_create_node_list(whole_nodes))
n2v_wholekg.add_edges_from(for_n2v_create_relation_list(whole_kg_data))
```

```python
rgcn_host_layerkg = nx.DiGraph()
rgcn_host_layerkg.add_nodes_from(for_rgcn_create_node_list(l1_nodes))
rgcn_host_layerkg.add_edges_from(for_rgcn_create_relation_list(full_L1))

rgcn_causal_layerkg = nx.DiGraph()
rgcn_causal_layerkg.add_nodes_from(for_rgcn_create_node_list(l2_nodes))
rgcn_causal_layerkg.add_edges_from(for_rgcn_create_relation_list(full_L2))

rgcn_amr_layerkg = nx.DiGraph()
rgcn_amr_layerkg.add_nodes_from(for_rgcn_create_node_list(l3_nodes))
rgcn_amr_layerkg.add_edges_from(for_rgcn_create_relation_list(full_L3))

rgcn_wholekg = nx.DiGraph()
rgcn_wholekg.add_nodes_from(for_rgcn_create_node_list(whole_nodes))
rgcn_wholekg.add_edges_from(for_rgcn_create_relation_list(whole_kg_data))
```

Preparing none edges for link prediction

Selected adge types:
- e/p to args
- ie/p to bacteria
- ie/p to gene
- disease to ie/p

```python
#order of nodes is preserved all the way into pyg HeteroData
node_list = list(n2v_wholekg.nodes)
np.savetxt('node_list.csv',node_list)
```

```python
ep_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 1}
arg_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 7}

full = list(nx.non_edges(rgcn_wholekg.subgraph(ep_nodes|arg_nodes).copy()))

ep_arg = random.sample(full, int(0.01*len(full)))

upper=[]
lower=[]
for i in ep_arg:
    if n2v_wholekg.nodes[i[0]]['node_type_index'] != n2v_wholekg.nodes[i[1]]['node_type_index']: #ensuring the relation is ep_arg
        upper.append(node_list.index(i[0]))
        lower.append(node_list.index(i[1]))
    else: 
        continue
ep_arg = [upper,lower]
```

```python
iep_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 2}
bact_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 8}

full = list(nx.non_edges(rgcn_wholekg.subgraph(iep_nodes|bact_nodes).copy()))

iep_bact = random.sample(full, int(0.01*len(full)))

upper=[]
lower=[]
for i in iep_bact:
    if n2v_wholekg.nodes[i[0]]['node_type_index'] != n2v_wholekg.nodes[i[1]]['node_type_index']:
        upper.append(node_list.index(i[0]))
        lower.append(node_list.index(i[1]))
    else: continue
iep_bact = [upper,lower]
```

```python
iep_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 2}
gene_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 5}

full = list(nx.non_edges(rgcn_wholekg.subgraph(iep_nodes|gene_nodes).copy()))

iep_gene = random.sample(full, int(0.01*len(full)))

upper=[]
lower=[]
for i in iep_gene:
    if n2v_wholekg.nodes[i[0]]['node_type_index'] != n2v_wholekg.nodes[i[1]]['node_type_index']:
        upper.append(node_list.index(i[0]))
        lower.append(node_list.index(i[1]))
    else: continue

iep_gene = [upper,lower]
```

```python
dis_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 0}
iep_nodes = {x for x,y in n2v_wholekg.nodes(data=True) if y['node_type_index'] == 2}


full = list(nx.non_edges(rgcn_wholekg.subgraph(dis_nodes|iep_nodes).copy()))

dis_iep = random.sample(full, int(0.01*len(full)))

upper=[]
lower=[]
for i in dis_iep:
    if n2v_wholekg.nodes[i[0]]['node_type_index'] != n2v_wholekg.nodes[i[1]]['node_type_index']:
            upper.append(node_list.index(i[0]))
            lower.append(node_list.index(i[1]))
    else: continue

dis_iep = [upper,lower]
```

```python
np.savetxt(f'n2v_rgcn_kg/lp_ep_arg.csv', ep_arg)

np.savetxt(f'n2v_rgcn_kg/lp_iep_bact.csv', iep_bact)

np.savetxt(f'n2v_rgcn_kg/lp_iep_gene.csv', iep_gene)

np.savetxt(f'n2v_rgcn_kg/lp_dis_iep.csv', dis_iep)
```
