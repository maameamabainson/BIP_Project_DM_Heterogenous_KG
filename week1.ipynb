## Data from [PrimeKG on Harvard Dataverse](https://dataverse.harvard.edu/dataset.xhtml?persistentId=doi:10.7910/DVN/IXA7BM)

```python
disease_features = pd.read_csv('dataverse_files/disease_features.csv')
primekg_nodes = pd.read_csv('dataverse_files/nodes.csv')
primekg_edges = pd.read_csv('dataverse_files/edges.csv')
primekg = pd.read_csv('dataverse_files/kg.csv')
```

```python
v1_t2dm = primekg[(primekg['x_name'] == 'type 2 diabetes mellitus') | (primekg['y_name'] == 'type 2 diabetes mellitus')] #Version 1 of T2DM  subgraph, relations where T2DM is either a source or recepient.

v1_t2dm = v1_t2dm[~(v1_t2dm['x_type'] == 'exposure') & ~(v1_t2dm['y_type'] == 'exposure')] #Ignoring 'exposure' nodes.
v1_t2dm.head(5)
```

```python
#According to the PrimeKG schema, there may be drug-phenotype, phenotype-gene, gene-pathway and gene-BP/MF/CC relations that lead to indirect connections to T2DM.

relevant_drugs1 = v1_t2dm[v1_t2dm['x_type'] == 'drug']['x_index'].values.tolist()
relevant_drugs2 = v1_t2dm[v1_t2dm['y_type'] == 'drug']['y_index'].values.tolist()
relevant_phenotypes1 = v1_t2dm[v1_t2dm['x_type'] == 'effect/phenotype']['x_index'].values.tolist()
relevant_phenotypes2 = v1_t2dm[v1_t2dm['y_type'] == 'effect/phenotype']['y_index'].values.tolist()
relevant_genes1 = v1_t2dm[v1_t2dm['x_type'] == 'gene/protein']['x_index'].values.tolist()
relevant_genes2 = v1_t2dm[v1_t2dm['y_type'] == 'gene/protein']['y_index'].values.tolist()
```

```python
relevant_drugs = list(set(relevant_drugs1 + relevant_drugs2))
relevant_phenotypes = list(set(relevant_phenotypes1 + relevant_phenotypes2))
relevant_genes = list(set(relevant_genes1 + relevant_genes2))
```

```python
# drug-phenotype

additional_relevant_drug_phenotype_edges1 = primekg[(primekg['x_index'].isin(relevant_drugs)) & (primekg['y_index'].isin(relevant_phenotypes))]
additional_relevant_drug_phenotype_edges2 = primekg[(primekg['x_index'].isin(relevant_phenotypes)) & (primekg['y_index'].isin(relevant_drugs))]
```

```python
additional_relevant_drug_phenotype_edges2 #Seems like there is no relevant edge between drugs and phenotype missed
```

```python
#phenotype-gene
additional_relevant_phenotype_gene_edges1 = primekg[(primekg['x_index'].isin(relevant_phenotypes)) & (primekg['y_index'].isin(relevant_genes))] 
additional_relevant_phenotype_gene_edges2 = primekg[(primekg['x_index'].isin(relevant_genes)) & (primekg['y_index'].isin(relevant_phenotypes))] 
```

```python
#gene-pathway/BP/MF/CC
additional_relevant_genes_pathway_BP_MF_CC_edges1 = primekg[(primekg['x_index'].isin(relevant_genes)) & (primekg['y_type'].isin(['pathway', 'biological_process', 'molecular_function', 'cellular_component']))]
additional_relevant_genes_pathway_BP_MF_CC_edges2 = primekg[(primekg['x_type'].isin(['pathway', 'biological_process', 'molecular_function', 'cellular_component'])) & (primekg['y_index'].isin(relevant_genes))] 
```

```python
v2_t2dm = pd.concat([v1_t2dm, additional_relevant_drug_phenotype_edges1, additional_relevant_drug_phenotype_edges2, additional_relevant_phenotype_gene_edges1, additional_relevant_phenotype_gene_edges2, additional_relevant_genes_pathway_BP_MF_CC_edges1, additional_relevant_genes_pathway_BP_MF_CC_edges2])
len(v2_t2dm)
```

Final T2DM dataset/subgraph of PrimeKG including indirect drugs-phenotype, phenotype-genes, and gene-pathway/biological process/molecular function/cellular compound relations.
